# Present Value

Fixed interest rate ```i```\
Definition with ```def``` function ```D(c1)```\
This functtion return the present value with ```return``` statement\



````python
i = 0.1

def D(c1):
  return c1/(1+i)

print(D(12.1))

print(D(11))
````

# Net Present Value
The net present value, is the sum of the cash outflow today and the present value of the cash inflow in one year.

````python
def NPV(c):
  return c[0] + D(c[1])

cA = (-10.5, 12.1)
cB = (-10.5, 11)

print(NPV(cA))
print(NPV(cB))
````

# Financial Assets
Financial assets are financial instruments (“contracts”) that have a fixed price today and an uncertain price in one year. Think of a share in the equity of a firm that conducts an investment project. Such a share might be available at a price today of . The price of the share in one year depends on the success of the investment project, i.e., whether a high cash inflow is observed in the state or a low one in the state.


````python
S0 = 10

S1 = np.array((12.5,7.5)) #the uncertain price in one year as a vector

#The first state
#The second state

S = (S0, S1)

S
````

# Risk

Risk in general refers to a situation in which a full probability distribution over future states of the economy is (assumed
to be) known.

````python
p = 0.4

1 - p
````

````python
P = np.array((p, 1-p))

P
````

# Expected Return

The **expected return** of a financial asset is given as
the expectation of the price in one year minus the price today.\
The **expected rate** of return then simply is the expected return divided by
the price today

````python
# Definition of expected return
def ER(x0, x1):
  return np.dot(P,x1) - x0
# The expected return for the previously defined financial asset
ER(S0, S1)
````

````python
# Definition of expected rate of return
def mu(x0, x1):
  return (np.dot(P,x1) - x0)/x0
# The expected rate of return calculated for that asset.
mu(S0, S1)
````

````python
def r(x0, x1):   #calculation of the rates of return vector.
  return (x1-x0)/x0

r(S0,S1)
````

````python
mu = np.dot(P, r(S0, S1))
mu
````

````python
def sigma2(P,r,mu):
  return np.dot(P, (r-mu)**2)

sigma2(P,r(S0,S1),mu) #The function applied to the rates of return vector.
````

````python
def sigma(P,r, mu):#The definition of the volatility.
  return np.sqrt(np.dot(P, (r-mu)**2))

sigma(P,r(S0,S1),mu) # applied to the rates of return vector.
````
