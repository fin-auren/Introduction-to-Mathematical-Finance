# Linear Algebra Review: Matrices, Eigenvalues, SVD, and Perron-Frobenius Theorem

## Lecture Overview

This lecture reviews the fundamental linear algebra concepts that will appear throughout the course:

1. Matrices
2. Eigenvalues and Eigenvectors
3. Diagonalization (Eigenvalue Decomposition)
4. Singular Value Decomposition (SVD)
5. Perron-Frobenius Theorem

The main idea is that a matrix can be viewed in two ways:

* As a collection of data
* As a linear transformation

Understanding both viewpoints is essential for applying linear algebra to finance, machine learning, statistics, and engineering.

---

# 1. Matrices

## Definition

A matrix is a rectangular arrangement of numbers.

Example:

$$A=\begin{bmatrix}
1 & 2 & 3 \\
2 & 3 & 4 \\
4 & 5 & 10
\end{bmatrix}
$$

Matrices may have any number of rows and columns.

---

## Matrix as Data

Example: Stock Price Dataset

Rows correspond to companies:

* Apple
* Morgan Stanley
* Google

Columns correspond to dates:

* July 1
* September 1
* October 1

Entries represent stock prices on those dates.

$$
A =
\begin{bmatrix}
400 & 500 & 600 \\
100 & 110 & 120 \\
5000 & 5200 & 5100
\end{bmatrix}
$$

Thus, matrices naturally represent datasets.

---

## Matrix as a Linear Transformation

An (m \times n) matrix defines a transformation:

$$
A : \mathbb{R}^n \rightarrow \mathbb{R}^m
$$

Example:

$$
A=
\begin{bmatrix}
2 & 0 \\
0 & 3
\end{bmatrix}
$$

Apply it to

$$
x=
\begin{bmatrix}
1\\
1
\end{bmatrix}
$$

Then

$$
Ax=
\begin{bmatrix}
2 \\
3
\end{bmatrix}
$$

Interpretation:

* x-coordinate scaled by 2
* y-coordinate scaled by 3

---

## Important Question

If a matrix is simply a dataset, why does it define a meaningful linear transformation?

This question motivates many techniques later in the course, especially eigenvalue decomposition and SVD.

---

# 2. Eigenvalues and Eigenvectors

## Definition

A scalar $(\lambda)$ and a nonzero vector $(v)$ are called an eigenvalue-eigenvector pair of a matrix (A) if

$$
Av= \lambda v
$$

Important:

* Eigenvalues and eigenvectors always come in pairs.
* Eigenvalues may be complex.

---

## Equivalent Form

Rearranging,

$$
(A-\lambda I)v=0
$$

A nonzero solution exists if and only if

$$
\det(A-\lambda I)=0
$$

---

## Characteristic Polynomial

For an $(n \times n)$ matrix,

$$
\det(A-\lambda I)
$$

is a polynomial of degree (n).

Since every polynomial has at least one complex root:

* Every square matrix has at least one eigenvalue.
* Every eigenvalue has a corresponding eigenvector.

---

# 3. Geometric Meaning of Eigenvectors

If

$$
Av=\lambda v
$$

then applying $(A)$ to $(v)$:

* does not change its direction
* only scales its magnitude

Thus eigenvectors are special directions preserved by the transformation.

---

## Why Eigenvectors Matter

For an arbitrary vector, a matrix may rotate, stretch, and distort it.

For an eigenvector, the matrix performs only a scaling operation.

Therefore eigenvectors provide the simplest way to understand a transformation.

---

# 4. Diagonalization

## Definition

A matrix is diagonalizable if

$$
A = UDU^{-1}
$$

where:

* $(U)$ is invertible
* $(D)$ is diagonal

---

## Diagonal Matrix

A diagonal matrix has nonzero entries only along its diagonal:

$$
D=
\begin{bmatrix}
\lambda_1 & 0 & 0\\
0 & \lambda_2 & 0\\
0 & 0 & \lambda_3
\end{bmatrix}
$$

---

## Geometric Interpretation

Suppose

$$
A = UDU^{-1}
$$

Then there exist eigenvectors

$$
v_1,v_2,v_3
$$

such that

$$
Av_1=\lambda_1v_1
$$

$$
Av_2=\lambda_2v_2
$$

$$
Av_3=\lambda_3v_3
$$

Thus the entire transformation can be understood by:

* Eigenvectors = special directions
* Eigenvalues = scaling factors

---

# 5. Symmetric Matrices

## Definition

A matrix is symmetric if

$$
A=A^T
$$

Example:

$$
\begin{bmatrix}
1 & 2\\
2 & 3
\end{bmatrix}
$$

---

## Key Theorems

### Theorem 1

Every real symmetric matrix is orthogonally diagonalizable.

$$
A = UDU^T
$$

where (U) is orthonormal.

---

### Theorem 2

Every real symmetric matrix has real eigenvalues.

---

## Why Symmetric Matrices Are Important

For symmetric matrices:

* Eigenvalues are real
* Eigenvectors are orthogonal
* Diagonalization always exists

Therefore symmetric matrices are much easier to analyze.

---

# 6. Limitations of Eigenvalue Decomposition

Eigenvalue decomposition has limitations:

1. Matrix must be square
2. Matrix may not be diagonalizable
3. Many real-world datasets are rectangular

We therefore need a more general decomposition.

---

# 7. Singular Value Decomposition (SVD)

## Main Theorem

For every $(m \times n)$ matrix $(A)$,

$$
A = U\Sigma V^T
$$

where:

* $(U)$ is orthonormal
* $(V)$ is orthonormal
* $(\Sigma)$ is diagonal

---

## Why SVD Is Important

### Eigenvalue Decomposition

Works only for:

* Square matrices
* Diagonalizable matrices

### Singular Value Decomposition

Works for:

* Square matrices
* Rectangular matrices
* Symmetric matrices
* Non-symmetric matrices
* Every matrix

---

# 8. Geometric Interpretation of SVD

SVD introduces two coordinate systems:

Input space:

$$
\mathbb{R}^n
$$

Output space:

$$
\mathbb{R}^m
$$

Input directions:

$$
v_1,v_2,\dots
$$

Output directions:

$$
u_1,u_2,\dots
$$

The transformation acts as

$$
Av_i=\sigma_i u_i
$$

where $(\sigma_i)$ are singular values.

---

## Comparison with Eigenvalue Decomposition

### Eigenvalue Decomposition

$$
Av_i=\lambda_i v_i
$$

Direction remains unchanged.

---

### SVD

$$
Av_i=\sigma_i u_i
$$

Direction may change.

A vector in the input space is mapped to a different direction in the output space.

---

# 9. Computing the SVD

## Step 1

Compute

$$
A^TA
$$

which is symmetric.

---

## Step 2

Find eigenvalues and eigenvectors of $(A^TA)$:

$$
\lambda_1,\lambda_2,\dots,\lambda_r
$$

with eigenvectors

$$
v_1,v_2,\dots,v_r
$$

---

## Step 3

Define singular values

$$
\sigma_i=\sqrt{\lambda_i}
$$

---

## Step 4

Compute left singular vectors

$$
u_i=\frac{Av_i}{\sigma_i}
$$

---

## Step 5

Construct

$$
V=[v_1,v_2,\dots]
$$

$$
U=[u_1,u_2,\dots]
$$

$$
\Sigma=
\begin{bmatrix}
\sigma_1 & & \\
& \sigma_2 & \\
&& \ddots
\end{bmatrix}
$$

Then

$$
A=U\Sigma V^T
$$

---

# 10. Reduced SVD

Singular values equal to zero contain no useful information.

Therefore:

* Remove zero singular values.
* Remove corresponding singular vectors.

This produces the reduced SVD.

---

## Advantages

For a stock-price matrix:

* 5 companies
* 365 days

The original matrix size is

$$
5 \times 365
$$

Reduced SVD stores only important directions, leading to:

* Data compression
* Lower memory usage
* Faster computations

---

# 11. Financial Interpretation of SVD

For stock-price data:

Rows = companies

Columns = dates

The singular vectors capture:

* Hidden factors
* Relationships among companies
* Correlation structures

Companies that move together often appear together in dominant singular vectors.

---

# 12. Perron-Frobenius Theorem

This theorem appears theoretical but has important applications in:

* Finance
* Probability
* Markov chains
* Network analysis

---

## Assumptions

Let (A) be:

* Square
* Symmetric
* Positive in every entry

---

## Theorem

### 1. Dominant Eigenvalue Exists

There exists an eigenvalue

$$
\lambda_0
$$

such that

$$
|\lambda| < \lambda_0
$$

for every other eigenvalue.

---

### 2. Positive Eigenvector Exists

The eigenvector corresponding to $(\lambda_0)$ has strictly positive entries.

$$
v=
\begin{bmatrix}
+\
+\
+
\end{bmatrix}
$$

---

### 3. Multiplicity One

$$
\lambda_0
$$

has multiplicity one.

Therefore:

* It is unique.
* Its positive eigenvector is unique (up to scaling).

---

# Intuition Behind Perron-Frobenius

Because every matrix entry is positive:

* Positive components reinforce each other.
* A dominant positive direction emerges.
* This direction corresponds to the largest eigenvalue.

---

# Why Perron-Frobenius Matters

Applications include:

* PageRank
* Markov chains
* Population models
* Network analysis
* Ross Recovery Theorem in finance

---

# Key Takeaways

### Matrices

* Represent both data and transformations.

### Eigenvalues and Eigenvectors

$$
Av=\lambda v
$$

* Reveal invariant directions.

### Symmetric Matrices

* Real eigenvalues
* Orthogonal eigenvectors
* Always diagonalizable

### Eigenvalue Decomposition

$$
A=UDU^{-1}
$$

* Breaks a transformation into scaling operations.

### Singular Value Decomposition

$$
A=U\Sigma V^T
$$

* Works for every matrix.
* Fundamental tool in data science and finance.

### Reduced SVD

* Keeps only important information.
* Enables compression and dimensionality reduction.

### Perron-Frobenius Theorem

For positive matrices:

* Unique dominant eigenvalue
* Positive eigenvector
* Strong theoretical and practical applications

# 2. Random Variables

A **random variable** assigns a numerical value to outcomes of a random experiment.

We distinguish between:

- **Discrete Random Variables**: denoted by \(X\)
- **Continuous Random Variables**: denoted by \(Y\)

---

## 2.1 Probability Mass Function (PMF)

A discrete random variable is characterized by its **probability mass function (PMF)**:

$$
f_X(x) = P(X=x)
$$

Properties:

$$
f_X(x) \ge 0
$$

and

$$
\sum_x f_X(x)=1
$$

### Example

Suppose

$$
P(X=1)=P(X=-1)=P(X=0)=\frac13
$$

Then

$$
f_X(1)=f_X(-1)=f_X(0)=\frac13
$$

---

## 2.2 Probability Density Function (PDF)

A continuous random variable is characterized by its **probability density function (PDF)**:

$$
f_Y(y)\ge 0
$$

and

$$
\int_{-\infty}^{\infty} f_Y(y)\,dy =1
$$

### Example: Uniform Distribution

If

$$
f_Y(y)=1,\qquad 0\le y\le 1
$$

then \(Y\) is uniformly distributed on \([0,1]\).

---

# 3. Probability and Expectation

## 3.1 Probability of an Event

### Discrete Case

For an event \(A\),

$$
P(A)=\sum_{x\in A} f_X(x)
$$

### Continuous Case

$$
P(A)=\int_A f_Y(y)\,dy
$$

---

## 3.2 Expectation (Mean)

### Discrete Random Variable

$$
E[X]=\sum_x x\,f_X(x)
$$

### Continuous Random Variable

$$
E[Y]=\int_{-\infty}^{\infty} y\,f_Y(y)\,dy
$$

---

# 4. Independence

Two random variables \(X_1\) and \(X_2\) are independent if

$$
P(X_1\in A,\;X_2\in B)
=
P(X_1\in A)\,P(X_2\in B)
$$

for every pair of events \(A,B\).

---

## Pairwise vs Mutual Independence

### Pairwise Independence

Every pair is independent:

$$
X_1 \perp X_2
$$

$$
X_1 \perp X_3
$$

$$
X_2 \perp X_3
$$

but all three together may fail to be independent.

### Mutual Independence

For every collection of events,

$$
P(X_1\in A_1,\dots,X_n\in A_n)
=
\prod_{i=1}^{n}
P(X_i\in A_i)
$$

This is the stronger condition and is usually assumed when we say random variables are independent.

---

# 5. Normal Distribution

The most important probability distribution in statistics is the **Normal Distribution**.

A random variable is said to be

$$
N(\mu,\sigma^2)
$$

if its density is

$$
f(x)
=
\frac{1}{\sigma\sqrt{2\pi}}
\exp\left(
-\frac{(x-\mu)^2}{2\sigma^2}
\right)
$$

for all real \(x\).

---

## Parameters

- Mean:

$$
E[X]=\mu
$$

- Variance:

$$
Var(X)=\sigma^2
$$

---

## Shape

The normal distribution:

- is bell-shaped
- is symmetric around \(\mu\)
- has thin tails
- appears naturally in many applications

---

# 6. Why Normal Distribution Is Not Ideal for Stock Prices

Suppose stock prices satisfy

$$
P_n-P_{n-1}
\sim
N(\mu,\sigma^2)
$$

This model assumes that absolute price changes are normally distributed.

### Problem

A \$1 move is treated the same whether the stock price is

- \$10
- \$1000

In reality, percentage changes matter more than absolute changes.

Hence a better assumption is:

$$
\frac{P_n-P_{n-1}}{P_{n-1}}
$$

or equivalently the logarithmic return is normally distributed.

This motivates the **Log-Normal Distribution**.

---

# 7. Log-Normal Distribution

A random variable \(X\) is log-normal if

$$
\ln(X)
$$

is normally distributed.

Formally,

$$
\ln(X)
\sim
N(\mu,\sigma^2)
$$

---

## Change of Variables Formula

If

$$
Y=\ln(X)
$$

then

$$
P(X\le x)
=
P(Y\le \ln x)
$$

Using the change-of-variable formula:

$$
f_X(x)
=
f_Y(\ln x)
\left|
\frac{d}{dx}\ln x
\right|
$$

Since

$$
\frac{d}{dx}\ln x
=
\frac1x
$$

we obtain

$$
f_X(x)
=
\frac{1}{x\sigma\sqrt{2\pi}}
\exp\left(
-\frac{(\ln x-\mu)^2}
{2\sigma^2}
\right)
$$

for

$$
x>0
$$

---

## Important Remark

For a normal distribution:

$$
\mu = \text{mean}
$$

$$
\sigma^2 = \text{variance}
$$

For a log-normal distribution, this is **not true**.

The parameters \(\mu\) and \(\sigma\) describe the normal distribution of

$$
\ln(X)
$$

not of \(X\) itself.

Thus:

$$
\mu \neq E[X]
$$

and

$$
\sigma^2 \neq Var(X)
$$

---

# 8. Exponential Family

Many important distributions belong to the **Exponential Family**.

A distribution belongs to this family if its density can be written as

$$
f(x;\theta)
=
h(x)c(\theta)
\exp
\left(
\sum_{i=1}^{k}
w_i(\theta)t_i(x)
\right)
$$

where:

- $h(x)$ depends only on $x$
- $c(\theta)$ depends only on parameters
- $w_i(\theta)$ depends only on parameters
- $t_i(x)$ depends only on $x$

---

## Examples

- Normal Distribution
- Log-Normal Distribution
- Poisson Distribution
- Exponential Distribution

---

## Importance

The exponential family has powerful statistical properties:

- convenient likelihood functions
- sufficient statistics
- easier inference
- closure under multiplication of densities for independent samples

---

# 9. Moments

To understand a random variable, we study its moments.

The \(k\)-th moment is

$$
E[X^k]
$$

Examples:

### First Moment

$$
E[X]
$$

(mean)

### Second Moment

$$
E[X^2]
$$

### Variance

$$
Var(X)
=
E[X^2]-E[X]^2
$$

---

# 10. Moment Generating Function (MGF)

The **Moment Generating Function** of \(X\) is

$$
M_X(t)
=
E[e^{tX}]
$$

---

## Important Remark

An MGF does **not always exist**.

Example:

- Log-normal distributions do not possess a finite MGF.

---

# 11. Why It Is Called a Moment Generating Function

Differentiate:

$$
M_X(t)
=
E[e^{tX}]
$$

The \(k\)-th derivative satisfies

$$
M_X^{(k)}(0)
=
E[X^k]
$$

Thus all moments can be extracted from the MGF.

---

## Taylor Expansion

The MGF can be written as

$$
M_X(t)
=
\sum_{k=0}^{\infty}
\frac{t^k}{k!}
E[X^k]
$$

Hence the MGF encodes all moments simultaneously.

---

# 12. Uniqueness Theorem

If two random variables have the same MGF:

$$
M_X(t)=M_Y(t)
$$

for all \(t\), then

$$
X
\text{ and }
Y
$$

have the same distribution.

Thus an MGF completely characterizes a distribution whenever it exists.

---

## Important Caveat

Having identical moments does **not** necessarily imply identical distributions.

Reason:

Two random variables may have

$$
E[X^k]=E[Y^k]
$$

for every \(k\),

yet neither MGF exists.

Therefore the uniqueness theorem cannot be applied.

---

# 13. Convergence Through MGFs

Suppose

$$
M_{X_n}(t)
\to
M_X(t)
$$

for every \(t\).

Then

$$
P(X_n\le x)
\to
P(X\le x)
$$

for all \(x\).

Hence convergence of MGFs implies convergence in distribution.

This fact is crucial in proving the Central Limit Theorem.

---

# 14. Law of Large Numbers (LLN)

Let

$$
X_1,\dots,X_n
$$

be IID random variables with

$$
E[X_i]=\mu
$$

and

$$
Var(X_i)=\sigma^2
$$

Define the sample mean

$$
\bar X_n
=
\frac1n
\sum_{i=1}^{n}
X_i
$$

---

## Theorem (Weak Law of Large Numbers)

For every

$$
\epsilon>0
$$

$$
P(|\bar X_n-\mu|>\epsilon)
\to 0
$$

as

$$
n\to\infty
$$

---

## Proof Sketch

### Mean

$$
E[\bar X_n]
=
\mu
$$

### Variance

Because of independence,

$$
Var(\bar X_n)
=
\frac{\sigma^2}{n}
$$

As \(n\) increases, variance shrinks.

Using Chebyshev's inequality:

$$
P(|\bar X_n-\mu|>\epsilon)
\le
\frac{Var(\bar X_n)}
{\epsilon^2}
$$

Therefore

$$
P(|\bar X_n-\mu|>\epsilon)
\le
\frac{\sigma^2}
{n\epsilon^2}
$$

and the right side tends to zero.

Hence

$$
\bar X_n
\to
\mu
$$

in probability.

---

# 15. Interpretation of LLN

The theorem explains why casinos make money.

Even if the casino has only a tiny edge:

$$
E[\text{casino profit}]>0
$$

a huge number of bets causes average profit to converge toward the expected profit.

Large variance may hide the edge in the short run, but over many trials the expectation dominates.

---

# 16. Central Limit Theorem (CLT)

The LLN says

$$
\bar X_n
\to
\mu
$$

But what happens to the fluctuations around \(\mu\)?

Consider

$$
Y_n
=
\sqrt{n}
(\bar X_n-\mu)
$$

---

## Theorem (Central Limit Theorem)

Let

$$
X_1,\dots,X_n
$$

be IID with

$$
E[X_i]=\mu
$$

and

$$
Var(X_i)=\sigma^2
$$

Then

$$
Y_n
=
\sqrt n(\bar X_n-\mu)
$$

converges in distribution to

$$
N(0,\sigma^2)
$$

That is,

$$
P(Y_n\le x)
\to
P(Z\le x)
$$

where

$$
Z\sim N(0,\sigma^2)
$$

---

## Significance

Regardless of the original distribution:

- Bernoulli
- Uniform
- Exponential
- Poisson

the normalized average approaches a normal distribution.

This explains why normal distributions appear so frequently in nature and finance.

---

# 17. Proof Idea of CLT Using MGFs

Assume MGFs exist.

The MGF of

$$
Y_n
=
\frac1{\sqrt n}
\sum_{i=1}^{n}
(X_i-\mu)
$$

is

$$
M_{Y_n}(t)
=
E[e^{tY_n}]
$$

Using independence:

$$
M_{Y_n}(t)
=
\left(
E
\left[
e^{\frac{t}{\sqrt n}(X-\mu)}
\right]
\right)^n
$$

Expand using Taylor series:

$$
e^x
=
1+x+\frac{x^2}{2!}
+\cdots
$$

Since

$$
E[X-\mu]=0
$$

the linear term vanishes.

The leading term becomes

$$
1+
\frac{t^2\sigma^2}{2n}
+o\left(\frac1n\right)
$$

Thus

$$
M_{Y_n}(t)
=
\left(
1+
\frac{t^2\sigma^2}{2n}
+o\left(\frac1n\right)
\right)^n
$$

Taking the limit:

$$
M_{Y_n}(t)
\to
\exp
\left(
\frac{t^2\sigma^2}{2}
\right)
$$

which is precisely the MGF of

$$
N(0,\sigma^2)
$$

Hence the CLT follows.

---

# 18. Statistical Estimation

Suppose a random variable has unknown mean

$$
\mu
$$

Take independent observations

$$
X_1,\dots,X_n
$$

and estimate the mean using

$$
\hat\mu
=
\frac1n
\sum_{i=1}^{n}
X_i
$$

---

## Why This Works

### Law of Large Numbers

Guarantees

$$
\hat\mu
\approx
\mu
$$

for large \(n\).

### Central Limit Theorem

Describes the distribution of the estimation error:

$$
\sqrt n(\hat\mu-\mu)
\Rightarrow
N(0,\sigma^2)
$$

Therefore:

- the estimator becomes increasingly accurate,
- errors become approximately normal,
- confidence intervals can be constructed.

---

# 19. Key Takeaways

1. A PMF describes discrete random variables; a PDF describes continuous random variables.
2. Normal distributions are fundamental but not ideal for modeling stock prices directly.
3. Stock prices are often modeled using log-normal distributions.
4. Many common distributions belong to the exponential family.
5. The Moment Generating Function encodes all moments of a distribution.
6. If MGFs coincide, distributions coincide.
7. The Law of Large Numbers states that sample averages converge to the true mean.
8. The Central Limit Theorem states that normalized averages converge to a normal distribution.
9. LLN explains long-run averages; CLT explains fluctuations around those averages.
10. Together, LLN and CLT form the foundation of statistical estimation and quantitative finance.

# Stochastic Processes, Random Walks, Markov Chains, and Martingales

## 1. Introduction to Stochastic Processes

A **stochastic process** is a collection of random variables indexed by time.

### Discrete-Time Stochastic Process

$$
X_0, X_1, X_2, \dots
$$

where time takes discrete values:

$$
t = 0,1,2,\dots
$$

### Continuous-Time Stochastic Process

$$
X_t,\quad t\ge 0
$$

where time is continuous.

---

## Alternative Definition

A stochastic process can also be viewed as:

> A probability distribution over a space of possible paths.

Instead of focusing on individual random variables, we think of the entire trajectory as one random object.

Each realization corresponds to one possible path.

This viewpoint is mathematically convenient because many properties of stochastic processes are naturally expressed in terms of paths.

---

# 2. Examples of Stochastic Processes

## Example 1: Deterministic Process

Define

$$
f(t)=t
$$

with probability 1.

There is only one possible path:

$$
(0,1,2,3,\dots)
$$

This is technically a stochastic process, although it contains no randomness.

---

## Example 2: Two Possible Paths

Define

$$
f(t)=t
$$

with probability

$$
\frac12
$$

and

$$
f(t)=-t
$$

with probability

$$
\frac12.
$$

Properties:

* Entire path is chosen at time 0.
* Once one value is known, all future values become known.
* Random variables are highly dependent.

---

## Example 3: Independent Choices at Every Time

For each time (t),

$$
f(t)=t
$$

or

$$
f(t)=-t
$$

independently with probability

$$
\frac12.
$$

Properties:

* Knowing past values gives almost no information about future values.
* Process oscillates between the two lines.
* Future behavior remains highly uncertain.

---

# 3. Main Questions in Stochastic Process Theory

When observing a process at time (t), we know the past but not the future.

Three fundamental questions arise:

---

## (A) Dependence Structure

Given the past,

$$
X_0,\dots,X_t,
$$

what can we infer about future values?

Example:

* Can past stock prices help predict future stock prices?

---

## (B) Long-Term Behavior

What happens as

$$
t\to\infty?
$$

Examples:

* Law of Large Numbers
* Central Limit Theorem
* Long-run averages

---

## (C) Boundary Events

How frequently do extreme events occur?

Examples:

* Probability of a stock dropping 10% for five consecutive days.
* Probability that a call center has more than 90% idle capacity.

---

# 4. Simple Random Walk

One of the most important stochastic processes.

---

## Definition

Let

$$
Y_1,Y_2,\dots
$$

be IID random variables satisfying

$$
P(Y_i=1)=\frac12
$$

and

$$
P(Y_i=-1)=\frac12.
$$

Define

$$
X_0=0
$$

and

$$
X_t=\sum_{i=1}^{t}Y_i.
$$

Then

$$
{X_t}_{t\ge0}
$$

is called a **one-dimensional simple random walk**.

---

## Interpretation

At each step:

* Move right with probability (1/2)
* Move left with probability (1/2)

Graphically:

$$
0 \to 1 \to 2 \to 1 \to 2 \to 3 \to \cdots
$$

or

$$
0 \to -1 \to 0 \to -1 \to -2 \to \cdots
$$

depending on the coin tosses.

---

# 5. Long-Term Behavior of a Random Walk

Recall:

$$
X_t=\sum_{i=1}^{t}Y_i.
$$

Since

$$
E[Y_i]=0
$$

and

$$
Var(Y_i)=1,
$$

we obtain

$$
E[X_t]=0
$$

and

$$
Var(X_t)=t.
$$

Hence

$$
\text{Std}(X_t)=\sqrt t.
$$

---

## Central Limit Theorem

The CLT gives

$$
\frac{X_t}{\sqrt t}
\Rightarrow
N(0,1).
$$

Therefore

$$
X_t
\approx
N(0,t)
$$

for large (t).

---

## Consequence

Most trajectories stay within the band

$$
-\sqrt t
\le
X_t
\le
\sqrt t.
$$

Although theoretically

$$
X_t
=\pm t
$$

is possible, such extreme behavior is extremely unlikely.

---

## Recurrence Property

A remarkable theorem states:

With probability 1,

$$
X_t=0
$$

occurs infinitely many times.

Thus the random walk keeps returning to the origin forever.

---

# 6. Important Properties of Random Walk

## Property 1: Mean Zero

$$
E[X_t]=0.
$$

---

## Property 2: Independent Increments

For times

$$
t_0 <t_1 < \cdots < t_k
$$

the increments

$$
X_{t_1}-X_{t_0},
\quad
X_{t_2}-X_{t_1},
\quad \dots
$$

are mutually independent.

Interpretation:

What happens between two non-overlapping intervals is independent.

---

## Property 3: Stationary Increments

For every

$$
h\ge0,
$$

the distribution of

$$
X_{t+h}-X_t
$$

is the same as the distribution of

$$
X_h.
$$

Interpretation:

Only the length of the interval matters, not its starting point.

---

# 7. Gambler's Ruin Problem

Suppose a player starts with balance

$$
0.
$$

At each round:

* Win $1 with probability (1/2)
* Lose $1 with probability (1/2)

Balance follows a simple random walk.

---

## Problem

Play until either:

$$
+100
$$

or

$$
-50
$$

is reached.

Find the probability of reaching

$$
+100
$$

before

$$
-50.
$$

---

## General Result

If barriers are

$$
+B
$$

and

$$
-A,
$$

then

$$
P(\text{hit }B\text{ first})
============================

\frac{A}{A+B}
$$

and

$$
P(\text{hit }-A\text{ first})
=============================

\frac{B}{A+B}.
$$

---

### Example

For

$$
A=50,\qquad B=100,
$$

we get

$$
P(\text{hit }100)
=================

# \frac{50}{150}

\frac13.
$$

Hence

$$
P(\text{hit }-50)
=================

\frac23.
$$

---

# 8. Markov Chains

A Markov chain is a stochastic process satisfying a special memory property.

---

## Key Idea

The future depends only on the present state.

The past is irrelevant once the current state is known.

---

## Formal Definition

A process is a Markov chain if

$$
P(X_{t+1}=s \mid X_0,\dots,X_t)
===============================

P(X_{t+1}=s \mid X_t)
$$

for every state (s).

---

## Interpretation

Knowledge of

$$
X_t
$$

summarizes all relevant information from the past.

---

# 9. Random Walk is a Markov Chain

For the random walk,

$$
P(X_{t+1}=X_t+1 \mid \text{history})
====================================

\frac12
$$

and

$$
P(X_{t+1}=X_t-1 \mid \text{history})
====================================

\frac12.
$$

Only the current position matters.

Therefore a simple random walk is a Markov chain.

---

# 10. Finite Markov Chains

Suppose the state space is finite:

$$
S={1,2,\dots,m}.
$$

Define

$$
P_{ij}
======

P(X_{t+1}=j\mid X_t=i).
$$

---

## Transition Matrix

Construct

$$
A=
\begin{bmatrix}
P_{11} & P_{12} & \cdots & P_{1m}\
P_{21} & P_{22} & \cdots & P_{2m}\
\vdots & \vdots & \ddots & \vdots\
P_{m1} & P_{m2} & \cdots & P_{mm}
\end{bmatrix}.
$$

Each row sums to 1:

$$
\sum_j P_{ij}=1.
$$

This matrix completely describes the Markov chain.

---

# 11. Multi-Step Transitions

Probability of moving from state (i) to state (j) in two steps:

$$
q_{ij}
======

\sum_k P_{ik}P_{kj}.
$$

This is precisely matrix multiplication.

Hence:

$$
Q=A^2.
$$

Similarly,

$$
A^n
$$

contains all (n)-step transition probabilities.

---

# 12. Example: Machine Reliability

States:

* Working (W)
* Broken (B)

Transition probabilities:

$$
A=
\begin{bmatrix}
0.99 & 0.01\
0.80 & 0.20
\end{bmatrix}.
$$

Interpretation:

* Working machine breaks with probability 0.01.
* Broken machine is repaired with probability 0.80.

---

# 13. Stationary Distribution

Suppose the long-run distribution is

$$
v=
\begin{bmatrix}
\pi_1\
\pi_2
\end{bmatrix}.
$$

Then

$$
Av=v.
$$

Thus (v) is an eigenvector corresponding to eigenvalue 1.

---

## Definition

A probability vector

$$
v=(\pi_1,\dots,\pi_m)
$$

satisfying

$$
Av=v
$$

is called a **stationary distribution**.

---

## Interpretation

If the chain starts with distribution (v), then after one step it still has distribution (v).

The distribution remains unchanged forever.

---

# 14. Perron-Frobenius Connection

If all entries of the transition matrix are positive, then:

1. Eigenvalue 1 exists.
2. It is the largest eigenvalue.
3. There is a unique positive eigenvector.
4. The chain converges to the stationary distribution.

Thus the long-run behavior is completely determined.

---

# 15. Martingales

Martingales model **fair games**.

---

## Definition

A stochastic process

$$
X_0,X_1,X_2,\dots
$$

is a martingale if

$$
E[X_{t+1}\mid X_0,\dots,X_t]
============================

X_t.
$$

---

## Interpretation

Given all current information,

the expected future value equals the present value.

There is no expected gain or loss.

---

# 16. Examples of Martingales

## Example 1: Random Walk

Since

$$
E[Y_{t+1}]=0,
$$

we obtain

$$
E[X_{t+1}\mid X_t]
==================

X_t.
$$

Therefore the random walk is a martingale.

---

## Example 2: Roulette Balance

Not a martingale.

Reason:

The casino has a positive edge.

Thus

$$
E[X_{t+1}\mid X_t]
<
X_t.
$$

---

## Example 3: Multiplicative Martingale

Let

$$
P(Y_i=2)=\frac13
$$

and

$$
P(Y_i=\frac12)=\frac23.
$$

Then

$$
E[Y_i]
======

2\cdot\frac13
+
\frac12\cdot\frac23
===================

1.

$$

Define

$$
X_0=1
$$

and

$$
X_k=\prod_{i=1}^{k}Y_i.
$$

Then

$$
E[X_{k+1}\mid X_k]
==================

# X_kE[Y_{k+1}]

X_k.
$$

Hence it is a martingale.

---

# 17. Stopping Times

A stopping time is a random time at which a decision to stop can be made using only present and past information.

---

## Definition

A nonnegative integer-valued random variable

$$
\tau
$$

is a stopping time if for every (k),

the event

$$
{\tau\le k}
$$

depends only on

$$
X_0,X_1,\dots,X_k.
$$

---

# 18. Examples of Stopping Times

## Example 1

First time balance reaches

$$
100.
$$

This is a stopping time.

---

## Example 2

First time balance reaches either

$$
100
$$

or

$$
-50.
$$

Also a stopping time.

---

## Example 3 (Not a Stopping Time)

First peak of the process.

To know whether time (t) is a peak, we must know

$$
X_{t+1}.
$$

Therefore future information is required.

Hence it is **not** a stopping time.

---

# 19. Optional Stopping Theorem

One of the most important martingale results.

---

## Theorem

Let

$$
{X_t}
$$

be a martingale and let

$$
\tau
$$

be a bounded stopping time.

Then

$$
E[X_\tau]
=========

E[X_0].
$$

---

## Interpretation

No stopping strategy can create a positive expected profit from a fair game.

A martingale remains fair even if we choose when to stop.

---

# 20. Application to Gambler's Ruin

Suppose:

* Start at 0.
* Stop at +100 or -50.

Let

$$
p=P(\text{hit }100\text{ first}).
$$

Since random walk is a martingale,

$$
E[X_\tau]=0.
$$

But

$$
X_\tau=
100
$$

with probability (p), and

$$
X_\tau=-50
$$

with probability (1-p).

Therefore

$$
100p-50(1-p)=0.
$$

Solving,

$$
150p=50
$$

gives

$$
p=\frac13.
$$

This agrees with the gambler's ruin formula.

---

# 21. Key Takeaways

1. A stochastic process is a collection of random variables indexed by time.
2. The simple random walk is a fundamental discrete-time stochastic process.
3. Random walks have:

   * Mean zero
   * Independent increments
   * Stationary increments
4. A Markov chain has the property that the future depends only on the present state.
5. Finite Markov chains are completely described by their transition matrices.
6. Long-run behavior is characterized by stationary distributions.
7. A martingale models a fair game.
8. Stopping times use only present and past information.
9. The Optional Stopping Theorem states that no stopping strategy can systematically beat a fair game.
10. Random walks are both Markov chains and martingales, making them central objects in probability theory and quantitative finance.

# Use Cases of Multiple Linear Regression in Finance

Multiple Linear Regression (MLR) is one of the most widely used statistical tools in finance. It helps quantify relationships between financial variables, make predictions, test economic theories, and support investment decisions.

## 1. Stock Return Prediction

One of the most common applications of regression in finance is predicting stock returns.

**Dependent Variable:**

* Future stock return

**Independent Variables:**

* Price-to-Earnings (P/E) ratio
* Market capitalization
* Book-to-Market ratio
* Momentum indicators
* Interest rates

Example:

$$
\text{Return} = \beta_0 + \beta_1(\text{P/E}) + \beta_2(\text{Market Cap}) + \beta_3(\text{Momentum}) + \epsilon
$$

Investors use the model to identify stocks expected to outperform the market.

---

## 2. Asset Pricing Models

Regression is fundamental in testing asset pricing theories.

### Capital Asset Pricing Model (CAPM)

$$
R_i - R_f = \alpha + \beta(R_m - R_f) + \epsilon
$$

Where:

* $(R_i)$ = Asset return
* $(R_f)$ = Risk-free rate
* $(R_m)$ = Market return

The estimated beta measures the asset's market risk.

### Multi-Factor Models

Examples:

* Fama-French Three-Factor Model
* Carhart Four-Factor Model

These models explain stock returns using multiple risk factors.

---

## 3. Portfolio Risk Analysis

Regression helps determine how different factors influence portfolio performance.

Factors may include:

* Market returns
* Interest rates
* Inflation
* Exchange rates
* Commodity prices

Portfolio managers use these relationships to understand sources of risk and return.

---

## 4. Economic Forecasting

Financial institutions use regression to forecast economic indicators.

Examples:

* GDP growth
* Inflation
* Unemployment rates
* Interest rates

These forecasts support:

* Investment decisions
* Monetary policy analysis
* Corporate planning

---

## 5. Bond Yield Modeling

Bond yields are often influenced by macroeconomic variables.

Example model:

$$
\text{Bond Yield} = \beta_0 + \beta_1(\text{Inflation}) + \beta_2(\text{GDP Growth}) + \beta_3(\text{Policy Rate}) + \epsilon
$$

Applications:

* Fixed-income valuation
* Interest-rate forecasting
* Risk management

---

## 6. Credit Risk Assessment

Banks use regression models to estimate default probabilities.

Independent variables may include:

* Income
* Debt-to-income ratio
* Credit score
* Loan amount
* Employment history

Applications:

* Loan approval
* Credit scoring
* Risk-based pricing

---

## 7. Volatility and Risk Modeling

Regression can identify factors affecting market volatility.

Examples:

* Economic announcements
* Interest rate changes
* Market sentiment indicators

Used for:

* Value-at-Risk (VaR)
* Stress testing
* Portfolio hedging

---

## 8. Time Series Forecasting

When observations are collected over time, regression models can include lagged variables.

Example:

$$
R_t = \beta_0 + \beta_1 R_{t-1} + \beta_2 R_{t-2} + \epsilon_t
$$

Applications:

* Stock return forecasting
* Exchange rate prediction
* Interest rate modeling

---

## 9. Factor Investing

Quantitative investment strategies often rely on regression.

Common factors:

* Value
* Momentum
* Quality
* Size
* Low volatility

Regression estimates the exposure of securities to these factors and helps construct factor-based portfolios.

---

## 10. Performance Attribution

Portfolio managers use regression to determine what drives portfolio returns.

Questions answered:

* How much return came from market exposure?
* How much came from sector selection?
* How much came from stock-picking skill?

This helps evaluate investment strategies and manager performance.

---

## Key Takeaway

In finance, Multiple Linear Regression is primarily used for:

1. **Prediction** – forecasting returns, prices, and economic variables.
2. **Risk Analysis** – measuring exposures to market and macroeconomic factors.
3. **Asset Pricing** – estimating factor sensitivities such as beta.
4. **Portfolio Management** – attribution, optimization, and factor investing.
5. **Credit and Risk Modeling** – assessing default and financial risk.

It forms the statistical foundation of many quantitative finance, risk management, and investment analysis techniques.
